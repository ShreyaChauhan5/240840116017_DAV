### **AIM:- Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets**

**Step - 1: Import libraries.**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

**Step - 2: Create dataset.**

```python
# Create dictionary to store sales information.
data = {
    "Month" : ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Units_sold" : [120, 135, 150, 145, 160, 175, 180, 170, 190, 210, 230, 250],
    "Revenue" : [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000,250000],
    "Profit" : [20000, 23000, 25000, 24000, 27000, 30000, 31000, 29000, 33000, 37000, 41000, 45000]
}

df = pd.DataFrame(data)
df
```

**Step - 3: Inspect the dataset.**

```python
# Display the first five rows
df.head()
```

```python
# Display the number of row and column.
df.shape
```

```python
# Display the names of all columns in dataset
df.columns
```

```python
# Display information about dataset.
df.info()
```

**Step - 4: Generate Descriptive Statistics**

```python
# Generate Descriptive Statistics for all numerical column
df.describe()
```

**Step - 5: Calculate mean**

```python
# Calculate the avg number of units sold per month.
average_units = df["Units_sold"].mean()
print("Average units sold :", average_units)
```

**Step - 6: Calculate median**




```python
# Calculate the middle value of the ubits_sold column.\
median_units = df["Units_sold"].median()
print("Median units sold:", median_units)
```

**Step - 7: Calculate max and min**

```python
# Find the number of units sold in a month
maximum_units = df["Units_sold"].max()
minimum_units = df["Units_sold"].min()
print("Maximum units sold:", maximum_units)
print("Minimum units sold:", minimum_units)
```

**Step - 8: Calculate the standard deviation**

```python
# Calculate the standard deviation of units sold.
std_units = df["Units_sold"].std()
print("Standard deviation of units sold:", std_units)
```

**Step - 9: Find the month with highest sales**

```python
# Find the index of the row having the maximum revenue
highest_sales_index = df["Revenue"].idxmax()

# Retrieve the month correspoding to that index
highest_sales_month = df.loc[highest_sales_index, "Month"]

# Retrieve the maximum revenue
highest_revenue = df.loc[highest_sales_index, "Revenue"]

# Print the results
print("Highest sales Month:", highest_sales_month)
print("Highest revenue:", highest_revenue)
```

**Step - 10: Find the month with the lowest sales.**

```python
# Find the index of the row having the minimum revenue
lowest_sales_index = df["Revenue"].idxmin()

# Retrieve the month correspoding to that index
lowest_sales_month = df.loc[lowest_sales_index, "Month"]

# Retrieve the minimum revenue
lowest_revenue = df.loc[lowest_sales_index, "Revenue"]

# Print the results
print("lowest sales Month:", lowest_sales_month)
print("lowest revenue:", lowest_revenue)
```

**Step - 11: Calculate total revenue.**

```python
# Calculate the total revenue generated during the year.
total_revenue = df["Revenue"].sum()
print("Total revenue generated:", total_revenue)
```

**Step - 12: Calculate total profit.**

```python
# Calculate total profit genereted during a year.
total_profit = df["Profit"].sum()
print("Total profit generated:", total_profit)
```

**Step - 13: Identify the sales trend.**

```python
# Create a linechart for monthly revenue
plt.figure(figsize=(10, 5))

# Plot month on the X-axis and revenue on the Y-axis.
plt.plot(df["Month"], df["Revenue"], marker='o')

# Add a title to the graph
plt.title("Monthly Revenue Trend")

# Add xlabel
plt.xlabel("Month")

#Add ylabel
plt.ylabel("Revenue")
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

**Step - 14: Create a linechart for monthly profit.**

```python
# Create a linechart for monthly profit
plt.figure(figsize=(10, 5))

# Plot month on the X-axis and Profit on the Y-axis.
plt.plot(df["Month"], df["Profit"], marker='o')

# Add a title to the graph
plt.title("Monthly Profit Trend")

# Add xlabel
plt.xlabel("Month")

#Add ylabel
plt.ylabel("Profit")
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

**Step - 15: Compare revenue and profit**

```python
# Create a line chart to compare revenue and profit
plt.figure(figsize=(10, 5))

# Plot Revenue
plt.plot(df["Month"], df["Revenue"], marker='o', label="Revenue")

# plot profit
plt.plot(df["Month"], df["Profit"], marker='o', label="Profit")

# Add a title to the graph
plt.title("Revenue vs Profit")

# Add xlabel
plt.xlabel("Month")

# Add Ylabel
plt.ylabel("Amount")
plt.xticks(rotation=45)

# Add legend
plt.legend()

# Display the graph
plt.show()
```

**Step - 16: Identify growth in sales**

```python
# Calculate the percentage change in the revenue from one month to next month.
df["Revenue_Growth_%"] = df["Revenue"].pct_change() * 100
df
```

**Step - 17 : identify months with increasing revenue**

```python
# filter the dataset to display only those months where the revenue growth is positive value.
increasing_sales = df[df["Revenue_Growth_%"] > 0]
df
```
