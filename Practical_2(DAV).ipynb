**Step-1: Import the pandas Library**

```python
import pandas as pd
```

**Step-2: Create sample data**

```python
# Create a dictionary to store student information.
data = {
    "ID": [101,102,103,104,105,106,107,108],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry","Aman"],
    "Age": [20, 21, None, 20, 20, 22, 23,20],
    "City": ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahmedabad"],
    "Marks": [85, None, 78, 88, 88, 81, 90, 93]
}
```

**Step-3: Create a dataframe**

```python
df = pd.DataFrame(data)    # Convert the dictionary into a Pandas DataFrame.
df                         # Display the DataFrame in tabular format.
```

**Step-4: Inspect the dataset.**

```python
# Display first five row of dataset.
df.head()
```

```python
# Display last five row of dataset.
df.tail()
```

```python
# Display detailed information about dataset.
df.info()
```

```python
# Generate statistical information.
df.describe()
```

```python
# Display dimension of dataset.
df.shape       # Output Format: (Number of rows, Number of columns)
```

```python
# Display the name of all columns.
df.columns
```

```python
# Display the data type of each column.
df.dtypes
```

**Step-5: Check missing values.**

```python
# Check for missing value(NULL) in every column.
df.isnull().sum()
```

**Step-6: Fill missing values.**

```python
# Fill missing marks.
# Replace missing value in marks column.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
# fillna() replace missing values with average.
```

```python
# Fill missing age.
# Using the average age of all student.
# Replace missing value in age column.

df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
# Fill missing names.
# Replace missing names with the text "Unknown".

df["Name"] = df["Name"].fillna("Unknown")
df
```

**Step-7: Detect duplicate records.**

```python
# Check whether each row is duplicate.
df.duplicated()

# True indicates a duplicate row.
# False indicates a unique row.
```

```python
# Count the total number of duplicate row.
df.duplicated().sum()
```

**Step-8: Remove duplicate records.**

```python
# Remove duplicate rows from dataframe.
df = df.drop_duplicates()          # removes repeated rows and retains only the first occurence.
df
```

**Step-9: Rename a column.**

```python
# Rename the column "Marks" to "Score".
df.rename(columns = {"Marks" : "Score"}, inplace=True)
df
```

**Srep-10: Change the data type.**

```python
# Convert each column from float to integer.
df["Age"] = df["Age"].astype(int)
df.dtypes
```

**Step-11: Create a new column.**

```python
# Create a new column named "Final Score".
df["Final Score"] = df["Score"] + 5         # Add 5 grace marks to every student's score.
df
```

**Step-12: Convert text to uppercase.**

```python
# Convert all student names to uppercase.
df["Name"] = df["Name"].str.upper()
df
```

**Step-13: Convert text to lowecase.**

```python
# Convert all city names to lowercase.
df["City"] = df["City"].str.lower()
df
```

Step-14: Remove extra spaces.

```python
# Remove leading and trailing spaces from city names.
df["City"] = df["City"].str.strip()
df
```

**Step-15: Sort the dataset.**

```python
# Sort the dataframe based on "Final Score" column.
df.sort_values(by= "Final Score", ascending=False)
```

**Step-16: Save the cleaned dataset.**

```python
# Save the cleaned dataframe as a new CSV file.
df.to_csv("cleaned_students.csv", index=False)
print("Dataset saved successfully.")
```
