### **AIM :- Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie)**

**Step-1 : Import important libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Step-2: Create Dataset.**

```python
#  Create Dataset to store sales data.
data = {

    "Month" : ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product" : ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region" : ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales" : [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold" : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(data)
df
```

**Step-3: Create a line chart.**

```python
# Create a line chart to see how did the company's sales changes from january to december.
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['Sales'], marker='o', linestyle='-', linewidth=2)
plt.title('Monthly Sales Trend (Jan - Dec)')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

Interpretation:            
As shown in chart, the company's sales generally increases from january to december, but in some months like april and august, sales decreases.

**Step-4: Create a bar chart.**

```python
# Create a bar chart to see which product generated how much total sales.
total_sales = df.groupby("Product")["Sales"].sum()
# Create bar chart
plt.figure(figsize=(8, 5))
total_sales.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")

# Show values on top of bars
for i, value in enumerate(total_sales):
    plt.text(i, value, f"{value:,}", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

Interpretation:                                            
From Chart, Laptop has generated total of 655,000 sales.                      
Mobile has generated total of 695,000 sales.                              
Tablet has generated total of 765,000 sales.

**Step-5: Create a Histogram.**

```python
# Create a histogram to see how are the individual sales values distributed.
plt.figure(figsize=(10, 6))
plt.hist(df['Sales'], bins=10, edgecolor='black')
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

Interpretation:                                                 
The lowest sales value is 120000 and highest sales value is 250000.            
Most sales values are concentrated between ₹130,000 and ₹190,000.

**Step-6: Create a scatter plot.**

```python
# Create a scatter plot to check if there is a relationship between the number of units sold and total sales.
plt.figure(figsize=(10, 6))
plt.scatter(df['Units_Sold'], df['Sales'])
plt.title('Units Sold vs Sales')
plt.xlabel('Units Sold')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

Interpretation:     
There is some variation in the points, so the relationship is not perfectly linear.      
The scatter plot shows a positive relationship between Units Sold and Sales.

**Step-7: Create a pie chart.**

```python
# Create a pie chart to see what percentage of total sales comes from each region.
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(6, 6))
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', startangle=130, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])
plt.title('Percentage of Total Sales by Region')
plt.show()
```

Interpretation:                       
West has the highest sales, approximately 26.7% of total sales.              
East has sales approximately 26.5%.           
South has sales approximately 24.6%.               
North has the lowest sales approximately 22.2%.
