### **AIM:- Apply filtering, sorting, aggregation, and GroupBy operations on real-world datasets.**

**Step-1: Import pandas library.**

```python
import pandas as pd
```

**Step-2: Create sample data.**

```python
# Create dictionary to store employee information.
data = {
    "Emp_ID" : [101,102,103,104,105,106,107,108],
    "Name" : ["Rahul", "Priya", "Amit", "Riya", "Karan", "Neha", "Vivek", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "HR"],
    "City" : ["Surat", "Navsari", "Surat", "Valsad", "Surat", "Navsari", "Valsad", "Surat"],
    "Salary" : [45000, 38000, 52000, 48000, 40000, 51000, 60000, 42000],
    "Experience" : [2, 1, 5, 4, 2, 6, 8, 3]
}
```

**Step-3: Create a dataframe.**

```python
df = pd.DataFrame(data)     # Convert dictionary into pandas a dataframe.
df                          # Display thedataframe into tabular format.
```

**Step-4: Display the dataset.**

```python
# Display the complete dataframe
df
```

**Step-5: filter data.**

```python
#  Select only employees working in the IT department.
df[df["Department"]=="IT"]
```

**Task-1: Display employees whose salary is greater than 50000.**

```python
# Display employees whose salary is greater than 50000.
df[df["Salary"]>50000]
```

**Task-2: Display employees from surat.**

```python
df[df["City"]=="Surat"]
```

**Task-3: Apply multiple conditions.**

```python
# Display employees from IT department.
# having salary greater than 50000.
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

**Step-6: Sort Data.**

**Task-1: Sort salary in ascending order.**

```python
# Sort employess by salary from lowest to highest.
df.sort_values(by="Salary")
```

**Task-2: Sort salary in decsending order.**

```python
# sort employees by salary from highest to lowest.
df.sort_values(by="Salary", ascending=False)
```

**Task-3: Sort by department and salary.**

```python
# first sort by department.
# within each department, sort salaries in descending order.
df.sort_values(by=["Department", "Salary"], ascending=[True,False])
```

**Step-7: Perform Aggregation.**

**Task-1: Calculate total salary.**

```python
# Calculate total salary.
df["Salary"].sum()
```

**Task-2: Calculate average salary.**

```python
# Calculate the avg salary.
df["Salary"].mean()
```

**Task-3: Find highest salary.**

```python
# Display the maximum salary.
df["Salary"].max()
```

**Task-4: Find lowest salary.**

```python
# Display the minimum salary.
df["Salary"].min()
```

**Task-5: Count Employees.**

```python
# Count the total number of employees.
df["Emp_ID"].count()
```

**Step-8: Perform GroupBy Operations.**

**Task-1: Average salary department-wise.**

```python
# Group employees based on department.
# Calculate the average salary of each department.
df.groupby("Department")["Salary"].mean()
```

**Task-2: Minimum salary department-wise.**

```python
# Group employees based on department.
# Calculate minimum salary of each department.
df.groupby("Department")["Salary"].min()
```

**Task-3: Count the total number of employees department-wise.**

```python
# Group employees based on department.
# Calculate total number of employees in each department.
df.groupby("Department")["Emp_ID"].count()
```

**Task-4: Total salary department-wise.**

```python
# Calculate the total salary paid by each department.
df.groupby("Department")["Salary"].sum()
```

**Task-5: Maximum salary department-wise.**

```python
# Calculate maximum salary of each department.
df.groupby("Department")["Salary"].max()
```

**Task-6: Display city-wise salary**

```python
# Group employees based on city
# Calculate average salary of employee from each city.
df.groupby("City")["Salary"].mean()
```

**Step-9: Multiple aggregation functions.**

```python
# Apply multiple statistocal functions
# to the salary column for each department.
df.groupby("Department")["Salary"].agg(["count","sum", "mean", "min", "max"])
```

**Step-10: Save the result.**

```python
# Save the dataset into a csv file.
df.to_csv("Employee_Data.csv", index=False)
print("Dataset saved successfully.")
```
