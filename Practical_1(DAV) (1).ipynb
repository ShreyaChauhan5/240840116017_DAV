**Step-1: Import the pandas library**

```python
# import pandas library.
import pandas as pd
```

**Step-2: Create sample data**

```python
# Create a dictionary to store student information.
data = {
    "ID": [101,102,103,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gadhinagar", "Ahmedabad"],
    "Marks": [85, 92, 78, 88, 81, 84, 90]
}
```

**Step-3: Create a dataframe**

```python
df = pd.DataFrame(data)    # Convert the dictionary into a Pandas DataFrame.
df                         # Display the DataFrame in tabular format.
```

**Step-4: Save data as a csv file**

```python
df.to_csv("students.csv", index=False)        # Save the DataFrame as a CSV file.
print("CSV file created successfully")        # Display confirmation after saving the file.
```

**Step-5: Save data as an excel file**

```python
df.to_excel("students.xlsx", index=False)        # Save the DataFrame as an Excel file.
print("Excel file created successfully")         # Display confirmation message.
```

**Step-6: Save data as a json file**

```python
df.to_json("students.json", orient="records", indent=4)  # Save the DataFrame as a formatted json file.
print("JSON file created successfully")                  # Display confirmation message.
```
